```python
import pandas as pd
# Task 1: Total runs by team
df = pd.DataFrame({
    'team': ['CSK', 'MI', 'CSK', 'MI', 'RCB'],
    'runs': [180, 160, 200, 170, 190]
})
print(df.groupby('team', as_index=False)['runs'].sum())
```

```python
# Task 2: Average order amount by city and restaurant
df = pd.DataFrame({
    'city': ['Ahmedabad', 'Ahmedabad', 'Mumbai', 'Mumbai'],
    'restaurant': ['A', 'A', 'B', 'B'],
    'order_amount': [300, 500, 400, 600]
})
print(df.groupby(['city', 'restaurant'])['order_amount'].agg('mean'))
```

```python
# Task 3: Total sales and average units sold
df = pd.DataFrame({
    'category': ['Mobile', 'Mobile', 'Laptop', 'Laptop'],
    'region': ['West', 'East', 'West', 'East'],
    'sales': [50000, 40000, 70000, 60000],
    'units_sold': [10, 8, 5, 6]
})
print(df.groupby(['category', 'region']).agg({
    'sales': 'sum',
    'units_sold': 'mean'
}))
```

```python
# Task 4: Revenue by region
df = pd.DataFrame({
    'region': ['West', 'East', 'West', 'East'],
    'revenue': [5000, 7000, 8000, 4000]
})
print(df.groupby('region').agg({
    'revenue': ['sum', 'max', 'min']
}))
```

```python
# Task 5: Total and average streams by genre
df = pd.DataFrame({
    'artist': ['A', 'B', 'C', 'D'],
    'genre': ['Pop', 'Rock', 'Pop', 'Rock'],
    'streams': [1000, 2000, 3000, 1500]
})
result = df.groupby('genre')['streams'].agg(
    total='sum',
    average='mean'
).sort_values('total', ascending=False)

print(result)
```
