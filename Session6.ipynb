```python
# Task 1
import pandas as pd

songs = pd.DataFrame({
    'song': ['Song A', 'Song B', 'Song A', 'Song C'],
    'artist': ['Arijit', 'Shreya', 'Arijit', 'Jubin'],
    'streams': [1000, 2000, 1000, 3000],
    'date': ['01-10-2026', '02-10-2026', '01-10-2026', '03-10-2026']
})

songs = songs.drop_duplicates(subset=['song'], keep='first')

print(songs)
```

```python
# Task 2

df = pd.DataFrame({
    'product_id': [1, 2, 1, 3],
    'user_id': [10, 20, 10, 30],
    'rating': [5, 4, 3, 5]
})

print("Before:", df.duplicated(['product_id', 'user_id']).sum())

df = df.drop_duplicates(['product_id', 'user_id'])

print("After:", df.duplicated(['product_id', 'user_id']).sum())
```

```python
# Task 3

df = pd.DataFrame({
    'order_id': [1, 2, 3, 4, 5],
    'order_amount': [100, 200, 300, 400, 2000]
})

Q1 = df['order_amount'].quantile(0.25)
Q3 = df['order_amount'].quantile(0.75)
IQR = Q3 - Q1

print(df[(df['order_amount'] < Q1 - 1.5 * IQR) |
         (df['order_amount'] > Q3 + 1.5 * IQR)])
```

```python
# Task 4

from scipy.stats import zscore

df = pd.DataFrame({
    'date': ['1 Oct', '2 Oct', '3 Oct', '4 Oct', '5 Oct'],
    'tickets_sold': [100, 110, 105, 120, 1000]
})

df['z_score'] = zscore(df['tickets_sold'])

print(df[df['z_score'].abs() > 2]['date'])
```

```python
# Task 5: Find outliers using Z-score
!pip install scipy
import pandas as pd
import numpy as np

def find_outliers(df):
    scores = zscore(df['amount'])
    return df[np.abs(scores) > 2]

wallet = pd.DataFrame({
    'transaction_id': [1, 2, 3, 4, 5, 6, 7],
    'amount': [100, 150, 120, 130, 110, 140, 2000]
})

result = find_outliers(wallet)

print(result)
```
