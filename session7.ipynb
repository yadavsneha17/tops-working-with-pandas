```python
#task 1
import pandas as pd
df = pd.DataFrame({
    'ProductName': ['Laptop', 'Mobile', 'Mouse'],
    'Price': [50000, 15000, 500],
    'Qty': [2, 3, 4]
})
df['TotalValue'] = df['Price'] * df['Qty']
print(df)
```

```python
#task 2
df = pd.DataFrame({
    'Item': ['Pizza', 'Burger', 'Pasta'],
    'Price': [200, 100, 150],
    'Qty': [2, 3, 1]
})
df['DiscountedPrice'] = df['Price'].apply(lambda x: x * 0.9)
print(df)
```

```python
#task 3
df = pd.DataFrame({
    'PostID': [1, 2, 3],
    'Likes': [1500, 2000000, 500],
    'Comments': [2000, 1500000, 300]
})
def format_num(x):
    if x >= 1000000:
        return str(round(x / 1000000, 1)) + 'M'
    elif x >= 1000:
        return str(round(x / 1000, 1)) + 'K'
    return str(x)
df[['Likes', 'Comments']] = df[['Likes', 'Comments']].applymap(format_num)
print(df)
```

```python
#task 4
df = pd.DataFrame({
    'Restaurant': ['Pizza Hut', 'Dominos', 'McDonalds'],
    'Price': [200, 300, 150],
    'Qty': [2, 1, 3],
    'DeliveryCharge': [30, 40, 20]
})
df['FinalAmount'] = df.apply(
    lambda x: x['Price'] * x['Qty'] + x['DeliveryCharge'],
    axis=1
)
print(df)
```

```python
#task 5
df = pd.DataFrame({
    'Product': ['Shirt', 'Shoes', 'Jeans'],
    'Price': [1000, 2000, 1500],
    'Qty': [2, 1, 2]
})
df['GST'] = df['Price'] * df['Qty'] * 0.05
print(df)
```
