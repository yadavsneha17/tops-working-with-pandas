```python
import pandas as pd
```

```python
# Task 1: Convert Instagram usernames to lowercase
df = pd.DataFrame({
    "username": ["CoolDude", "foodieQueen", "SnehaYadav", "TechLover", "PrettyGirl"]
})

df["username"] = df["username"].str.lower()
print("Task 1: Lowercase usernames")
print(df)
```

```python
# Task 2: Remove Remix from song titles
songs = pd.Series([
    "tum hi ho bandhu Remix",
    "tere sang yara Remix",
    "Heeriye Remix",
    "piya ghar aaya"
])

songs = songs.str.replace("Remix", "", regex=False).str.strip()
print("\nTask 2: Cleaned song titles")
print(songs)

```

```python
# Task 3: Filter reviews containing delivery
reviews = pd.DataFrame({
    "review": [
        "Fast delivery",
        "Food was tasty",
        "Late Delivery",
        "Good service",
        "Excellent delivery"
    ]
})

result = reviews[reviews["review"].str.contains("delivery", case=False, na=False)]
print("\nTask 3: Delivery reviews")
print(result)
```

```python
# Task 4: Extract main product categories
categories = pd.Series([
    "Electronics/Mobiles/Smartphones",
    "Home/Kitchen/Appliances",
    "Fashion/Men/Shirts",
    "Beauty/Skincare/Creams"
])

df = pd.DataFrame({"category": categories})
df["main_category"] = df["category"].str.split("/").str[0]

print("\nTask 4: Main product categories")
print(df)
```

```python
# Task 5: Convert titles to uppercase and replace spaces
videos = pd.DataFrame({
    "title": [
        "Python Tutorial",
        "Data Analytics",
        "Power BI Dashboard"
    ]
})

videos["title"] = videos["title"].str.upper().str.replace(" ", "_", regex=False)

print("\nTask 5: YouTube video titles")
print(videos)
```
