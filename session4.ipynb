```python
import pandas as pd
```

```python
# Task 1: Sort Bollywood movies by rating
movies = pd.DataFrame({
    "title": ["Dangal", "3 Idiots", "Pathaan", "Jawan", "12th Fail", "Sholay"],
    "rating": [8.3, 8.4, 5.8, 6.9, 8.7, 8.1],
    "release_year": [2016, 2009, 2023, 2023, 2023, 1975]
})

print("Task 1: Movies sorted by rating")
print(movies.sort_values("rating", ascending=False))
```

```python
# Task 2: Sort IPL teams by index
teams = pd.DataFrame({
    "team": ["CSK", "MI", "RCB", "KKR", "GT"],
    "points": [14, 16, 12, 18, 10],
    "net_run_rate": [0.5, 0.8, 0.2, 1.1, -0.3]
})

print("\nTask 2: IPL teams sorted by index")
print(teams.sort_index(ascending=False))
```

```python
# Task 3: Reset YouTube videos index
videos = pd.DataFrame({
    "video_title": ["Python Tutorial", "Data Analytics", "Power BI"],
    "views": [10000, 25000, 18000],
    "likes": [500, 1200, 900]
}, index=[5, 8, 12])

videos = videos.reset_index(drop=True)

print("\nTask 3: YouTube videos with reset index")
print(videos)
```

```python
# Task 4: Reindex Zomato restaurants
restaurants = pd.DataFrame({
    "restaurant": ["Dominos", "McDonalds", "La Pinoz", "Subway"],
    "rating": [4.2, 4.0, 4.3, 3.9]
}, index=["A", "B", "C", "D"])

restaurants = restaurants.reindex(["C", "A", "B", "D"])

print("\nTask 4: Reindexed restaurants")
print(restaurants)
```

```python
# Task 5: Sort products by discount and reset index
products = pd.DataFrame({
    "product_name": ["Laptop", "Mobile", "Headphones", "Smartwatch"],
    "price": [50000, 20000, 3000, 5000],
    "discount": [10, 25, 40, 15]
})

products = products.sort_values("discount", ascending=False)
products = products.reset_index(drop=True)

print("\nTask 5: Products sorted by discount")
print(products)
```
