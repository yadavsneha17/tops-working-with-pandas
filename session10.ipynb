```python
import pandas as pd
```

```python
# Task 1: Create users DataFrame
users = pd.DataFrame({
    'user_id': [1, 2, 3, 4, 5],
    'username': ['Sneha', 'Priya', 'Riya', 'Neha', 'Pooja'],
    'city': ['Ahmedabad', 'Surat', 'Mumbai', 'Delhi', 'Pune']
})
# Create orders DataFrame
orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105, 106, 107],
    'user_id': [1, 2, 1, 3, 5, 2, 3],
    'order_amount': [500, 300, 700, 250, 900, 400, 600]
})
# Inner Join
result1 = pd.merge(users, orders, on='user_id', how='inner')
print("Inner Join:")
print(result1)
```

```python
# Task 2: Left Join
result2 = pd.merge(users, orders, on='user_id', how='left')
print("\nLeft Join:")
print(result2)
```

```python
# Task 3
restaurants = pd.DataFrame({
    'restaurant_id': [1, 2, 3],
    'name': ['Food Hub', 'Pizza Point', 'Burger House'],
    'city': ['Ahmedabad', 'Surat', 'Mumbai']
})
zomato_reviews = pd.DataFrame({
    'review_id': [101, 102, 103],
    'restaurant_id': [1, 2, 4],
    'rating': [4.5, 4.0, 3.5],
    'reviewer_city': ['Ahmedabad', 'Surat', 'Delhi']
})
result = pd.merge(
    restaurants,
    zomato_reviews,
    on='restaurant_id',
    how='outer'
)
print(result)
```

```python
# Task 4
playlists = pd.DataFrame({
    'playlist_id': [101, 102, 103, 104],
    'user_id': [1, 2, 3, 4],
    'city': ['Ahmedabad', 'Surat', 'Mumbai', 'Delhi'],
    'genre': ['Pop', 'Rock', 'Jazz', 'Classical']
})
user_profiles = pd.DataFrame({
    'user_id': [1, 2, 3, 4],
    'username': ['Sneha', 'Priya', 'Riya', 'Neha'],
    'city': ['Ahmedabad', 'Mumbai', 'Mumbai', 'Delhi']
})
result = pd.merge(
    playlists,
    user_profiles,
    on=['user_id', 'city'],
    how='inner'
)
print(result)
```

```python
# Task 5
products = pd.DataFrame({
    'product_id': [1, 2, 3],
    'name': ['Laptop', 'Mobile', 'Headphones'],
    'category': ['Electronics', 'Electronics', 'Accessories']
})
ratings = pd.DataFrame({
    'rating_id': [101, 102, 103, 104],
    'product_id': [1, 2, 3, 4],
    'rating': [4.5, 4.0, 5.0, 3.5]
})
result = pd.merge(
    products,
    ratings,
    on='product_id',
    how='right'
)
print(result)
```
