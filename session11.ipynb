```python
import pandas as pd
```

```python
# Task 1: Combine January and February orders
jan = pd.DataFrame({"Order": [101, 102], "Amount": [200, 300]})
feb = pd.DataFrame({"Order": [103, 104], "Amount": [400, 500]})
print(pd.concat([jan, feb], ignore_index=True))
```

```python
# Task 2: Combine product details and ratings
products = pd.DataFrame({"Product": ["Mobile", "Laptop"]})
ratings = pd.DataFrame({"Rating": [4.5, 4.2]})
print(pd.concat([products, ratings], axis=1))
```

```python
# Task 3: Pivot Spotify streams
df = pd.DataFrame({
    "user": ["A", "A", "B", "B", "C", "C"],
    "day": ["Mon", "Tue", "Mon", "Tue", "Mon", "Tue"],
    "streams": [100, 120, 90, 110, 80, 95]
})
print(df.pivot(index="user", columns="day", values="streams"))
```

```python
# Task 4: Average restaurant orders per month
df = pd.DataFrame({
    "restaurant": ["A", "A", "B", "B"],
    "month": ["Jan", "Feb", "Jan", "Feb"],
    "total_amount": [200, 300, 400, 500]
})
print(df.pivot_table(index="restaurant", columns="month",
                     values="total_amount", aggfunc="mean"))
```

```python
# Task 5: Combine monthly sales and calculate total
jan = pd.DataFrame({"product": ["Shirt", "Shoes"], "sales": [1000, 2000]})
feb = pd.DataFrame({"product": ["Shirt", "Shoes"], "sales": [1500, 2500]})
mar = pd.DataFrame({"product": ["Shirt", "Shoes"], "sales": [1200, 3000]})

sales = pd.concat([jan, feb, mar], ignore_index=True)
print(sales.groupby("product")["sales"].sum())
```
