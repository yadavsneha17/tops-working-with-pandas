```python
import pandas as pd

# Task 1: Load dataset
df = pd.read_csv('/ipl matches.csv')
print(df.head(10))

# Task 2: Univariate Analysis
print("Mean:", df['win_by_runs'].mean())
print("Median:", df['win_by_runs'].median())
print("Mode:", df['win_by_runs'].mode().tolist())
print("Minimum:", df['win_by_runs'].min())
print("Maximum:", df['win_by_runs'].max())

# Task 3: Zomato Correlation
zomato = pd.DataFrame({
    'name': ['A', 'B', 'C', 'D', 'E'],
    'rating': [4.2, 4.5, 3.9, 4.7, 4.1],
    'votes': [120, 200, 80, 250, 100],
    'avg_cost': [300, 500, 250, 600, 350]
})
print("Correlation:", zomato['rating'].corr(zomato['votes']))

# Task 4: Summary Statistics
print(df.describe())
print("Insight: describe() shows the average and highest runs.")

# Task 5: Additional Analysis
# Analysis 1: Runs and Strike Rate
# Analysis 2: Average Runs per Match
df['runs_per_match'] = df['win_by_runs']
print(df[['runs_per_match']].head(10))
```
