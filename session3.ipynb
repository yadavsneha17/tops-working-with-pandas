```python
# Task 1

import pandas as pd
movies = pd.DataFrame({
    'title': ['Pathaan', 'Jawan', 'Dangal', '3 Idiots'],
    'genre': ['Action', 'Action', 'Drama', 'Comedy'],
    'rating': [7.5, 7.0, 8.3, 8.4],
    'box_office': [1000, 1100, 2000, 500]
})

print(movies.loc[:, ['title', 'rating']])
```

```python
# Task 2

products = pd.DataFrame({
    'product_name': ['Phone', 'Laptop', 'Watch', 'Shoes', 'Bag',
                     'Tablet', 'Headphones', 'Camera', 'Keyboard', 'Mouse'],
    'category': ['Electronics', 'Electronics', 'Accessories', 'Fashion',
                 'Fashion', 'Electronics', 'Electronics', 'Electronics',
                 'Electronics', 'Electronics'],
    'price': [15000, 50000, 2000, 1500, 1200, 18000, 2500, 30000, 800, 500]
})

print(products.iloc[:10, :])
```

```python
# Task 3

orders = pd.DataFrame({
    'restaurant': ['Cafe A', 'Cafe B', 'Cafe C', 'Cafe D'],
    'order_amount': [450, 700, 850, 300],
    'delivery_time': [25, 35, 30, 20]
})

print(orders[orders['order_amount'] > 500])
```

```python
# Task 4

songs = pd.DataFrame({
    'song': ['Song A', 'Song B', 'Song C', 'Song D'],
    'artist': ['A', 'B', 'C', 'D'],
    'streams': [1500000, 800000, 2000000, 1200000],
    'duration': [170, 160, 200, 150]
})

result = songs.query('streams > 1000000 and duration < 180')

print(result)
```

```python
# Task 5

ipl_df = pd.DataFrame({
    'team1': ['CSK', 'MI', 'RCB', 'GT', 'KKR'],
    'team2': ['MI', 'RCB', 'GT', 'CSK', 'SRH'],
    'winner': ['CSK', 'MI', 'GT', 'CSK', 'KKR'],
    'total_runs': [190, 175, 220, 230, 205]
})

result = ipl_df[
    (ipl_df['total_runs'] >= 180) &
    (ipl_df['total_runs'] <= 220)
]

print(result)
```
