```python
import pandas as pd
```

```python
# Task 1: Convert date strings to datetime
dates = pd.Series([
    "15-01-2026",
    "20-02-2026",
    "10-03-2026",
    "25-04-2026",
    "05-05-2026"
])

dates = pd.to_datetime(dates, format="%d-%m-%Y")
print("Task 1: Datetime dates")
print(dates)
```

```python
# Task 2: Extract year, month and day
movies = pd.DataFrame({
    "movie": ["Dangal", "Jawan", "Pathaan", "Sholay"],
    "release_date": ["23-12-2016", "07-09-2023",
                     "25-01-2023", "15-08-1975"]
})

movies["release_date"] = pd.to_datetime(
    movies["release_date"], format="%d-%m-%Y"
)

movies["year"] = movies["release_date"].dt.year
movies["month"] = movies["release_date"].dt.month
movies["day"] = movies["release_date"].dt.day

print("\nTask 2: Movie release dates")
print(movies)
```

```python
# Task 3
print(pd.date_range(start=pd.Timestamp.today().normalize(), periods=7))
```

```python
# Task 4
df = pd.DataFrame({
    "user_id": [101, 102, 103],
    "start_date": ["01-01-2026", "15-03-2026", "01-06-2026"]
})
df["start_date"] = pd.to_datetime(df["start_date"], format="%d-%m-%Y")
df["tenure_days"] = (pd.Timestamp.today().normalize() - df["start_date"]).dt.days
print(df)
```
