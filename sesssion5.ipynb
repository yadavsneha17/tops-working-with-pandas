```python
#task 1

import pandas as pd

ipl_df = pd.DataFrame({
    'PlayerName': ['Virat Kohli', 'Rohit Sharma', 'MS Dhoni', 'Shubman Gill'],
    'Team': ['RCB', 'MI', 'CSK', 'GT'],
    'Age': [37, None, 44, None]
})

print(ipl_df[ipl_df['Age'].isnull()])
```

```python
# Task 2

orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104],
    'customer_name': ['Aman', 'Riya', 'Neha', 'Raj'],
    'delivery_rating': [5, None, 4, None]
})

cleaned_df = orders.dropna(subset=['delivery_rating'])

print(cleaned_df)
```

```python
# Task 3

products = pd.DataFrame({
    'Product': ['Phone', 'Laptop', 'Watch', 'Headphones'],
    'rating': [4.0, 5.0, None, 3.0]
})

products['rating'] = products['rating'].fillna(
    products['rating'].mean()
)

print(products)
```

```python
# Task 4

songs = pd.DataFrame({
    'Song': ['Song A', 'Song B', 'Song C', 'Song D'],
    'duration': [180, 200, None, 240]
})

songs['duration'] = songs['duration'].fillna(
    songs['duration'].median()
)

songs.to_csv('cleaned_spotify.csv', index=False)

print(songs)
```

```python
# Task 5

users = pd.DataFrame({
    'Username': ['sneha', 'riya', 'aman', 'neha'],
    'followers': [1200, None, 850, None]
})

users['followers'] = users['followers'].fillna(0)

print(users)
```
