```python
import pandas as pd
```

```python
# Task 1: Favorite Cricket Teams
fav_cricket_teams = pd.Series(["CSK", "MI", "RCB", "KKR", "GT"])
print("Task 1: Favorite Cricket Teams")
print(fav_cricket_teams)
```

```python
# Task 2: Food Orders
food_orders = pd.DataFrame({
    "Item": ["Pizza", "Burger", "Sandwich"],
    "Restaurant": ["Dominos", "McDonalds", "Subway"],
    "Price": [250, 150, 120]
})
print("\nTask 2: Food Orders")
print(food_orders)
```

```python
# Task 3: Instagram Followers
followers = pd.Series([1200, 54000, 780, 150000, 32000])
print("\nTask 3: Highest Followers")
print(followers.idxmax(), followers.max())
print("Lowest Followers")
print(followers.idxmin(), followers.min())
```

```python
# Task 4: Music Playlist
playlist = pd.DataFrame({
    "Song": ["Tum Hi Ho", "Kesariya", "Heeriye", "Apna Bana Le"],
    "Artist": ["Arijit Singh", "Arijit Singh", "Jasleen Royal", "Arijit Singh"],
    "Duration": [4.2, 4.3, 3.3, 4.0]
})
print("\nTask 4: Songs and Duration")
print(playlist[["Song", "Duration"]])
```
